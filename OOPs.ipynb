```python
class Student:
    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks

    def greet(self):
        return f"Hello, my name is {self.name} and I am {self.age} years old."
    def get_grade(self):
        if self.marks >= 90:
            return "A"
        elif self.marks >= 80:
            return "B"
        elif self.marks >= 70:
            return "C"
        elif self.marks >= 60:
            return "D"
        else:
            return "F"
    def is_passed(self):
        return self.marks >= 60
s1 = Student("Payal", 20, 85)
print(s1.greet())
print("Grade:", s1.get_grade())
print("Passed:", s1.is_passed())
s2 = Student("John", 22, 55)
print(s2.greet())
print("Grade:", s2.get_grade())
print("Passed:", s2.is_passed())
s3 = Student("Smita", 19, 95)
print(s3.greet())
print("Grade:", s3.get_grade())
print("Passed:", s3.is_passed())
```

```python
class Student:
    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.__marks = marks  # Private variable

    def greet(self):
        return f"Hello, my name is {self.name} and I am {self.age} years old."

    def get_grade(self):
        if self.__marks >= 90:
            return "A"
        elif self.__marks >= 80:
            return "B"
        elif self.__marks >= 70:
            return "C"
        elif self.__marks >= 60:
            return "D"
        else:
            return "F"

    def is_passed(self):
        return self.__marks >= 60

    def show_marks(self):
        print("Marks:", self.__marks)

    def update_marks(self, new_marks):
        self.__marks = new_marks
        print("Marks updated successfully!")


# Student 1
s1 = Student("Payal", 20, 85)

print(s1.greet())
print("Grade:", s1.get_grade())
print("Passed:", s1.is_passed())
s1.show_marks()

# Update marks using method
s1.update_marks(100)
s1.show_marks()

# Try modifying marks directly
s1.__marks = 50
print("Direct access:", s1.__marks)

# Original private variable remains unchanged
s1.show_marks()

# Student 2
s2 = Student("John", 22, 55)
print("\n" + s2.greet())
print("Grade:", s2.get_grade())
print("Passed:", s2.is_passed())
s2.show_marks()

# Student 3
s3 = Student("Smita", 19, 95)
print("\n" + s3.greet())
print("Grade:", s3.get_grade())
print("Passed:", s3.is_passed())
s3.show_marks()
```
