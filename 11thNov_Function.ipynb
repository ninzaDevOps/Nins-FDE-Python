Python Function: with parameter and return value
Accept input argument from the function call


```python
def payal(a,b):
    return a+b
result=payal(10,20)
print(result)
```

```python
def calculate_bill(price,quantity,discount):
    total=price*quantity
    discount_amt=total*discount/100
    final_amt=total-discount_amt 
    return final_amt
bill=calculate_bill(1000,10,20)
print("Final Bill",bill)

```

```python
def calculate_bill(price,quantity,discount):
    total=price*quantity
    discount_amt=total*discount/100
    final_amt=total-discount_amt 
    return final_amt , discount_amt, total
calculate_bill(1000,10,20)

```
